```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = ['Січ', 'Лют', 'Бер', 'Кві', 'Тра', 'Чер', 'Лип', 'Сер', 'Вер', 'Жов', 'Лис', 'Гру']
plan = [120, 130, 140, 150, 160, 170, 165, 180, 190, 200, 210, 230]
fact = [115, 135, 138, 155, 162, 168, 175, 185, 180, 205, 215, 240]

plt.figure(figsize=(10, 5))

plt.plot(months, plan, label='План', marker='o', linestyle='--', color='#2b5c8f', linewidth=2)
plt.plot(months, fact, label='Факт', marker='s', linestyle='-', color='#d95f02', linewidth=2)

plt.title('Порівняння планових та фактичних продажів за місяцями', fontsize=13)
plt.xlabel('Місяць', fontsize=11)
plt.ylabel('Обсяг продажів (тис. грн)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left', frameon=True)

plt.tight_layout()
plt.show()
```

```python
np.random.seed(42)
ages = np.random.randint(18, 71, size=100)
mean_age = np.mean(ages)

plt.figure(figsize=(8, 5))

plt.hist(ages, bins=10, color='#4575b4', edgecolor='black', alpha=0.8)

plt.axvline(
    mean_age, 
    color='red', 
    linestyle='dashed', 
    linewidth=2, 
    label=f'Середній вік: {mean_age:.1f} р.'
)

plt.title('Гістограма розподілу віку вибірки (100 осіб)', fontsize=13)
plt.xlabel('Вік (років)', fontsize=11)
plt.ylabel('Кількість осіб', fontsize=11)
plt.legend(loc='upper right')
plt.grid(axis='y', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
np.random.seed(42)
group_a = np.clip(np.random.normal(loc=74, scale=12, size=35), 0, 100)
group_b = np.clip(np.random.normal(loc=83, scale=8, size=30), 0, 100)
group_c = np.clip(np.random.normal(loc=69, scale=15, size=32), 0, 100)

plt.figure(figsize=(8, 5))

plt.boxplot(
    [group_a, group_b, group_c], 
    tick_labels=['Група A', 'Група B', 'Група C'],
    patch_artist=True,
    boxprops=dict(facecolor='#a6cee3', color='#1f78b4'),
    medianprops=dict(color='darkred', linewidth=2)
)

plt.title('Порівняння результатів іспитів за групами', fontsize=13)
plt.xlabel('Академічна група', fontsize=11)
plt.ylabel('Підсумковий бал', fontsize=11)
plt.grid(axis='y', linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
days = ['Пн, 01.11', 'Вт, 02.11', 'Ср, 03.11', 'Чт, 04.11', 'Пт, 05.11', 'Сб, 06.11', 'Нд, 07.11']
temperature = [11, 13, 10, 14, 16, 15, 12]  # у °C
humidity = [65, 70, 78, 62, 58, 64, 72]     # у %

fig, ax1 = plt.subplots(figsize=(9, 5))

color_temp = '#e34a33'
line1 = ax1.plot(days, temperature, color=color_temp, marker='o', linewidth=2, label='Температура (°C)')
ax1.set_xlabel('Дата', fontsize=11)
ax1.set_ylabel('Температура (°C)', color=color_temp, fontsize=11)
ax1.tick_params(axis='y', labelcolor=color_temp)
ax1.tick_params(axis='x', rotation=45)

ax2 = ax1.twinx()
color_hum = '#2b8cbe'
line2 = ax2.plot(days, humidity, color=color_hum, marker='s', linestyle='--', linewidth=2, label='Вологість (%)')
ax2.set_ylabel('Вологість (%)', color=color_hum, fontsize=11)
ax2.tick_params(axis='y', labelcolor=color_hum)

lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Показники температури та вологості за тиждень', fontsize=13)
plt.grid(True, linestyle=':', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
hours = list(range(24))
load = [18, 14, 11, 9, 10, 16, 28, 52, 70, 82, 85, 80, 78, 84, 88, 93, 89, 74, 65, 58, 48, 38, 29, 21]

plt.figure(figsize=(10, 5))

plt.plot(hours, load, color='#1f77b4', linewidth=2.5, label='Навантаження CPU')

plt.fill_between(hours, load, color='#1f77b4', alpha=0.25)

plt.title('Погодинне навантаження сервера протягом доби', fontsize=13)
plt.xlabel('Година доби (0–23)', fontsize=11)
plt.ylabel('Рівень навантаження (%)', fontsize=11)

plt.xticks(range(0, 24, 2))
plt.ylim(0, 105)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
timeline = [f'Тижд {i}' for i in range(1, 9)]

conversion = [2.1, 2.3, 2.5, 2.4, 2.8, 3.0, 3.1, 3.4]      # %
retention = [42, 44, 45, 47, 49, 50, 48, 52]              # %
avg_check = [420, 435, 410, 460, 475, 470, 490, 510]       # грн
orders_count = [1150, 1220, 1280, 1260, 1390, 1450, 1510, 1620]

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

axes[0, 0].plot(timeline, conversion, color='#8856a7', marker='o', linewidth=2)
axes[0, 0].set_title('1. Конверсія (%)', fontsize=11, fontweight='bold')
axes[0, 0].grid(True, linestyle=':', alpha=0.6)

axes[0, 1].plot(timeline, retention, color='#1c9099', marker='s', linewidth=2)
axes[0, 1].set_title('2. Утримання користувачів (%)', fontsize=11, fontweight='bold')
axes[0, 1].grid(True, linestyle=':', alpha=0.6)

axes[1, 0].plot(timeline, avg_check, color='#31a354', marker='^', linewidth=2)
axes[1, 0].set_title('3. Середній чек (грн)', fontsize=11, fontweight='bold')
axes[1, 0].grid(True, linestyle=':', alpha=0.6)

axes[1, 1].plot(timeline, orders_count, color='#e6550d', marker='d', linewidth=2)
axes[1, 1].set_title('4. Кількість замовлень (шт.)', fontsize=11, fontweight='bold')
axes[1, 1].grid(True, linestyle=':', alpha=0.
for ax in axes.flat:
    ax.tick_params(axis='x', rotation=30)

fig.suptitle('Огляд ключових продуктових метрик за 8 тижнів', fontsize=15, y=0.99)

plt.tight_layout()
plt.show()
```
